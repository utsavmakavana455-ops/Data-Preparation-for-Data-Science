```python
import pandas as pd
df = pd.read_csv('netflix_titles.csv')


#Convert date_added to Datetime
df["date_added"] = pd.to_datetime(df["date_added"], errors="coerce")

print(df["date_added"].head())
```

```python
#Extract year_added
df["year_added"] = df["date_added"].dt.year
print(df["year_added"].head())
```

```python
# Extract Numeric Duration
df["duration_num"] = df["duration"].str.extract(r"(\d+)").astype(float)


# Extract Duration Type
df["duration_type"] = df["duration"].str.extract(r"(min|Season|Seasons)")

print(df[["duration", "duration_num", "duration_type"]].head())
```

```python
df["type_encoded"] = df["type"].map({
    "Movie": 0,
    "TV Show": 1
})

print(df[["type", "type_encoded"]].head())
```

```python
#Content Age = Current Year - Release Year


current_year = 2026

df["content_age"] = current_year - df["release_year"]
print(df["content_age"].head())
```

```python
df["num_genres"] = df["listed_in"].fillna("").apply(
    lambda x: len([g for g in x.split(",") if g.strip()])
)

print(df["num_genres"].head())
```
